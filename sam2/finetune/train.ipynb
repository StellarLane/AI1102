```python
import numpy as np
import torch
import cv2
import os
from sam2.build_sam import build_sam2
from sam2.sam2_image_predictor import SAM2ImagePredictor
from sam2.automatic_mask_generator import SAM2AutomaticMaskGenerator
from train_utils import *
from sklearn.metrics import jaccard_score

%matplotlib inline
%load_ext autoreload
%autoreload 2
```

```python
predictor.model.sam_mask_decoder.train(True) # enable training of mask decoder
optimizer=torch.optim.AdamW(params=predictor.model.parameters(),lr=1e-5,weight_decay=4e-5)
scaler = torch.amp.GradScaler('cuda')
losses = []
for itr in range(1000):
    with torch.amp.autocast('cuda'): # cast to mix precision
            image, mask, input_point, input_label = load_one_data("../datasets/train/") # load data batch
            image = image.copy()
            mask = mask.copy()
            if mask.shape[0]==0: continue # ignore empty batches
            predictor.set_image(image) # apply SAM image encoder to the image

            # prompt encoding

            mask_input, unnorm_coords, labels, unnorm_box = predictor._prep_prompts(input_point, input_label, box=None, mask_logits=None, normalize_coords=True)
            sparse_embeddings, dense_embeddings = predictor.model.sam_prompt_encoder(points=(unnorm_coords, labels),boxes=None,masks=None,)

            # mask decoder

            batched_mode = unnorm_coords.shape[0] > 1 # multi object prediction
            high_res_features = [feat_level[-1].unsqueeze(0) for feat_level in predictor._features["high_res_feats"]]
            low_res_masks, prd_scores, _, _ = predictor.model.sam_mask_decoder(image_embeddings=predictor._features["image_embed"][-1].unsqueeze(0),image_pe=predictor.model.sam_prompt_encoder.get_dense_pe(),sparse_prompt_embeddings=sparse_embeddings,dense_prompt_embeddings=dense_embeddings,multimask_output=True,repeat_image=batched_mode,high_res_features=high_res_features,)
            prd_masks = predictor._transforms.postprocess_masks(low_res_masks, predictor._orig_hw[-1])# Upscale the masks to the original image resolution

            # Segmentaion Loss caclulation

            gt_mask = torch.tensor(mask.astype(np.float32)).cuda()
            prd_mask = torch.sigmoid(prd_masks[:, 0])# Turn logit map to probability map
            seg_loss = (-gt_mask * torch.log(prd_mask + 0.00001) - (1 - gt_mask) * torch.log((1 - prd_mask) + 0.00001)).mean() # cross entropy loss

            # Score loss calculation (intersection over union) IOU

            inter = (gt_mask * (prd_mask > 0.5)).sum(1).sum(1)
            iou = inter / (gt_mask.sum(1).sum(1) + (prd_mask > 0.5).sum(1).sum(1) - inter)
            score_loss = torch.abs(prd_scores[:, 0] - iou).mean()
            loss=seg_loss+score_loss*0.05  # mix losses

            # apply back propogation

            predictor.model.zero_grad() # empty gradient
            scaler.scale(loss).backward()  # Backpropogate
            scaler.step(optimizer)
            scaler.update() # Mix precision

            if itr%100==0: torch.save(predictor.model, "fined_tuned_1.pt")

            if itr==0: mean_iou=0
            mean_iou = mean_iou * 0.99 + 0.01 * np.mean(iou.cpu().detach().numpy())
            losses.append(mean_iou)
            print("step)",itr, "Accuracy(IOU)=",mean_iou)
            if itr % 50 == 0:  # Display prediction results every 50 steps
                plt.figure(figsize=(12, 4))
                
                # plt.subplot(1, 3, 1)
                # plt.imshow(image)
                # plt.title(f'Original Image (Step {itr})')
                # plt.axis('off')
                
                # plt.subplot(1, 3, 2)
                # plt.imshow(mask[0], cmap='gray')
                # plt.title('Ground Truth Mask')
                # plt.axis('off')
                
                # plt.subplot(1, 3, 3)
                plt.imshow(prd_mask[0].cpu().detach().numpy(), cmap='gray')
                plt.title(f'Predicted Mask (IOU: {iou[0]:.3f})')
                plt.axis('off')
                print(itr)
                # plt.tight_layout()
                plt.show()
            if itr % 100 == 0 and len(losses) > 0:
                plt.figure(figsize=(8, 4))
                plt.plot(losses)
                plt.xlabel('Iteration')
                plt.ylabel('Mean IOU')
                plt.title('IOU over Training')
                plt.grid(True)
                plt.show()
```

```python
predictor.model.sam_mask_decoder.train(False) # enable training of mask decoder
predictor.model.sam_prompt_encoder.train(True)
optimizer=torch.optim.AdamW(params=predictor.model.parameters(),lr=1e-5,weight_decay=4e-5)
scaler = torch.amp.GradScaler('cuda')

losses = []
for itr in range(1000):
    with torch.amp.autocast('cuda'): # cast to mix precision
            image, mask, input_point, input_label = load_one_data("./dataset_0424/train/") # load data batch
            image = image.copy()
            mask = mask.copy()
            if mask.shape[0]==0: continue # ignore empty batches
            predictor.set_image(image) # apply SAM image encoder to the image

            # prompt encoding

            mask_input, unnorm_coords, labels, unnorm_box = predictor._prep_prompts(input_point, input_label, box=None, mask_logits=None, normalize_coords=True)
            sparse_embeddings, dense_embeddings = predictor.model.sam_prompt_encoder(points=(unnorm_coords, labels),boxes=None,masks=None,)

            # mask decoder

            batched_mode = unnorm_coords.shape[0] > 1 # multi object prediction
            high_res_features = [feat_level[-1].unsqueeze(0) for feat_level in predictor._features["high_res_feats"]]
            low_res_masks, prd_scores, _, _ = predictor.model.sam_mask_decoder(image_embeddings=predictor._features["image_embed"][-1].unsqueeze(0),image_pe=predictor.model.sam_prompt_encoder.get_dense_pe(),sparse_prompt_embeddings=sparse_embeddings,dense_prompt_embeddings=dense_embeddings,multimask_output=True,repeat_image=batched_mode,high_res_features=high_res_features,)
            prd_masks = predictor._transforms.postprocess_masks(low_res_masks, predictor._orig_hw[-1])# Upscale the masks to the original image resolution

            # Segmentaion Loss caclulation

            gt_mask = torch.tensor(mask.astype(np.float32)).cuda()
            prd_mask = torch.sigmoid(prd_masks[:, 0])# Turn logit map to probability map
            seg_loss = (-gt_mask * torch.log(prd_mask + 0.00001) - (1 - gt_mask) * torch.log((1 - prd_mask) + 0.00001)).mean() # cross entropy loss

            # Score loss calculation (intersection over union) IOU

            inter = (gt_mask * (prd_mask > 0.5)).sum(1).sum(1)
            iou = inter / (gt_mask.sum(1).sum(1) + (prd_mask > 0.5).sum(1).sum(1) - inter)
            score_loss = torch.abs(prd_scores[:, 0] - iou).mean()
            loss=seg_loss+score_loss*0.05  # mix losses

            # apply back propogation

            predictor.model.zero_grad() # empty gradient
            scaler.scale(loss).backward()  # Backpropogate
            scaler.step(optimizer)
            scaler.update() # Mix precision

            if itr%1==0: torch.save(predictor.model.state_dict(), "fined_tuned_2.pt")

            if itr==0: mean_iou=0
            mean_iou = mean_iou * 0.99 + 0.01 * np.mean(iou.cpu().detach().numpy())
            losses.append(mean_iou)
            print("step)",itr, "Accuracy(IOU)=",mean_iou)
```

```python
# Load the base model first, then load the fine-tuned weights
sam2_model_finetuned = build_sam2(model_cfg, sam2_checkpoint, device="cuda") # load base model
sam2_model_finetuned = torch.load("./fined_tuned_1.pt", weights_only=False) # load fine tuned model
sam2_checkpoint = "../checkpoints/sam2_hiera_tiny.pt" # path to model weight (pre model loaded from: https://dl.fbaipublicfiles.com/segment_anything_2/072824/sam2_hiera_small.pt)
model_cfg = "configs/sam2/sam2_hiera_t.yaml" #  model config
sam2_model = build_sam2(model_cfg, sam2_checkpoint, device="cuda") # load model
predictor_finetuned = SAM2ImagePredictor(sam2_model_finetuned)
predictor = SAM2ImagePredictor(sam2_model)
# Test and compare the two models on the validation dataset
test_results = []
num_test_samples = 60
predictor.model.sam_mask_decoder.train(True) # enable training of mask decoder
def get_one_mask(predictor = predictor, image=None, input_point=None, input_label=None, gt_mask=None):
    predictor.set_image(image) # apply SAM image encoder to the image
    mask_input, unnorm_coords, labels, unnorm_box = predictor._prep_prompts(input_point, input_label, box=None, mask_logits=None, normalize_coords=True)
    sparse_embeddings, dense_embeddings = predictor.model.sam_prompt_encoder(points=(unnorm_coords, labels), boxes=None, masks=None)

    batched_mode = unnorm_coords.shape[0] > 1 # multi object prediction
    high_res_features = [feat_level[-1].unsqueeze(0) for feat_level in predictor._features["high_res_feats"]]
    low_res_masks, prd_scores, _, _ = predictor.model.sam_mask_decoder(
            image_embeddings=predictor._features["image_embed"][-1].unsqueeze(0),
            image_pe=predictor.model.sam_prompt_encoder.get_dense_pe(),
            sparse_prompt_embeddings=sparse_embeddings,
            dense_prompt_embeddings=dense_embeddings,
            multimask_output=True,
            repeat_image=batched_mode,
            high_res_features=high_res_features
    )
    prd_masks = predictor._transforms.postprocess_masks(low_res_masks, predictor._orig_hw[-1]) # Upscale the masks to the original image resolution
    
    # Convert predicted mask to numpy and apply threshold
    prd_mask_sigmoid = torch.sigmoid(prd_masks[:, 0])
    prd_mask_np = (prd_mask_sigmoid[0].cpu().detach().numpy() > 0.5).astype(np.uint8)
    
    # Convert ground truth mask to numpy
    if torch.is_tensor(gt_mask):
        gt_mask_np = gt_mask[0].cpu().detach().numpy().astype(np.uint8)
    else:
        gt_mask_np = gt_mask[0].astype(np.uint8)
    
    # Resize predicted mask to match ground truth if needed
    if prd_mask_np.shape != gt_mask_np.shape:
        prd_mask_np = cv2.resize(prd_mask_np, (gt_mask_np.shape[1], gt_mask_np.shape[0]), interpolation=cv2.INTER_NEAREST)
    
    # Calculate IOU using jaccard_score with properly shaped arrays
    iou = jaccard_score(gt_mask_np.flatten(), prd_mask_np.flatten(), average='binary')
    
    return prd_mask_sigmoid, iou


print("Testing and comparing models on validation dataset...")
for i in range(num_test_samples):
        image, mask, input_point, input_label = load_one_data("../datasets/val/") # load data batch
        image = image.copy()
        mask = mask.copy()
        if mask.shape[0]==0:
                continue # ignore empty batches
        prd_mask, iou = get_one_mask(predictor, image, input_point, input_label, mask)
        prd_mask_finetune, iou_finetune = get_one_mask(predictor_finetuned, image, input_point, input_label, mask)
        test_results.append({
            'original_iou': iou,
            'finetuned_iou': iou_finetune,
            'improvement': iou_finetune - iou
        })

        # Visualize results every 10 samples
        if i % 10 == 0:
            plt.figure(figsize=(15, 5))
            
            # plt.subplot(1, 4, 1)
            # plt.imshow(image)
            # plt.title('Original Image')
            # plt.axis('off')
            
            # plt.subplot(1, 4, 2)
            # plt.imshow(mask[0], cmap='gray')
            # plt.title('Ground Truth')
            # plt.axis('off')
            
            # plt.subplot(1, 4, 3)
            plt.imshow(prd_mask[0].cpu().detach().numpy(), cmap='gray')
            plt.title(f'Original Model (IOU: {iou:.3f})')
            plt.axis('off')
            
            # plt.subplot(1, 4, 4)
            # plt.imshow(prd_mask_finetune[0].cpu().detach().numpy(), cmap='gray')
            # plt.title(f'Fine-tuned Model (IOU: {iou_finetune:.3f})')
            # plt.axis('off')
            
            # plt.tight_layout()
            # plt.show()
        print(iou_finetune - iou)
```
